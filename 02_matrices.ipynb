```python
import numpy as np

trips = np.array([
    [2, 5, 1],
    [0, 4, 3],
    [6, 0, 0]
])

prices = np.array([1, 0.5, 2])

np.dot(trips, prices)
```

```python
import numpy as np

trips = np.array([
    [2, 5, 1],
    [0, 4, 3],
    [6, 0, 0]
])

shopX = np.array([1, 0.5, 2])
shopY = np.array([2, 1, 1])

shopX_total = np.dot(trips, shopX)
shopY_total = np.dot(trips, shopY)

print(shopX_total, shopY_total)
```
